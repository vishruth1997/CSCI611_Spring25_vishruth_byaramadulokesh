### Testing on inferences and storing them in Detection folder

```python
import os
import cv2
import matplotlib.pyplot as plt

%matplotlib inline

test_images_folder = "/Users/akshayaralikatti/Downloads/Course_Work_25-Spring/611-AML/assignments/Yolo/AK_try/US-Road-Signs-58/test/images"
output_folder = "./detections"

os.makedirs(output_folder, exist_ok=True)

for image_name in sorted(os.listdir(test_images_folder)):
    if image_name.endswith((".jpg", ".png", ".jpeg")):
        image_path = os.path.join(test_images_folder, image_name)

        #running the inference
        results = model(image_path, conf=0.5, iou=0.4)

        #save detected image
        for result in results:
            output_image = result.plot()
            output_path = os.path.join(output_folder, f"detected_{image_name}")
            cv2.imwrite(output_path, output_image)

print("Inference completed on all test images")

#To display detected images

```

```python
def show_detected_images(folder_path, num_images=16, start_index=None):
    # Get the list of image files
    image_files = sorted(os.listdir(folder_path))
    
    if not image_files:
        print("No detected images found")
        return 

    # Calculate the start index for the middle images if not provided
    if start_index is None:
        total_images = len(image_files)
        start_index = max(0, (total_images - num_images) // 2)  # Middle index

    # Slice the list to get the desired set of images
    image_files = image_files[start_index:start_index + num_images]

    rows, cols = 4, 4
    fig, axes = plt.subplots(rows, cols, figsize=(15, 15))

    # Flatten axes for easier indexing
    axes = axes.flatten()

    for i, img_file in enumerate(image_files):
        img_path = os.path.join(folder_path, img_file)
        print(f"Loading: {img_path}")
        img = cv2.imread(img_path)

        if img is None:
            print(f"Warning: Failed to load {img_file}")
            continue

        # Convert BGR to RGB for displaying
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        # Display the image
        axes[i].imshow(img)
        axes[i].axis("off")
        axes[i].set_title(img_file, fontsize=10)

    # Hide unused subplots if there are fewer images than the grid size
    for j in range(len(image_files), len(axes)):
        axes[j].axis("off")

    plt.tight_layout()
    plt.show()

# Example usage
output_folder = "/Users/akshayaralikatti/Downloads/Course_Work_25-Spring/611-AML/assignments/Yolo/detections"
show_detected_images(output_folder, num_images=16)  # Automatically gets the middle 16 images
```
